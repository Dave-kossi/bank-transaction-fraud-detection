```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go

# Charger le jeu de données
data = pd.read_csv('Fraud_dataset.csv')

# Afficher les premières lignes du jeu de données
#print(data.head())

# Afficher les informations sur le jeu de données 
#print(data.info())

#Affichage des statistiques sur les colonnes des fraudes label [0,1] 

fig = px.pie(
    data, 
    names='Fraud_Label', 
    title='Distribution des Fraudes',
    color='Fraud_Label',
    color_discrete_map={0: '#66b3ff', 1: '#ff9999'},
    labels={'Fraud_Label': 'Type'},
    hole=0.3,
    height=600,
    width=600
)

fig.update_traces(
    textposition='inside',
    textinfo='percent+label',
    hovertemplate='<b>%{label}</b><br>Count: %{value}<br>Percentage: %{percent}<br><extra></extra>'
)

fig.update_layout(
    title_x=0.5,
    showlegend=False,
    font=dict(size=12)
)

fig.show()


```
